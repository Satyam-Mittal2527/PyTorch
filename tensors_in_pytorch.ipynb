## tensor Operations using GPU

```python
import torch
torch.cuda.is_available()
```

```python
device = torch.device('cuda')
```

```python
# creating a new tensor on GPU
torch.rand((2,3), device=device)
```

```python
# moving tensors in GPU
a = torch.rand(2,3)
b = a.to(device)
```

```python
import torch
import time

size = 10000

# Create matrices on CPU
matrix_cpu1 = torch.rand(size, size)
matrix_cpu2 = torch.rand(size, size)

# CPU computation
start_time = time.time()
result_cpu = torch.matmul(matrix_cpu1, matrix_cpu2)
cpu_time = time.time() - start_time

print(f"Time on CPU: {cpu_time:.4f} seconds")


# Move matrices to GPU
matrix_gpu1 = matrix_cpu1.to('cuda')
matrix_gpu2 = matrix_cpu2.to('cuda')

# GPU computation
torch.cuda.synchronize()  # Ensure GPU is ready
start_time = time.time()

result_gpu = torch.matmul(matrix_gpu1, matrix_gpu2)

torch.cuda.synchronize()  # Wait for GPU computation to finish
gpu_time = time.time() - start_time

print(f"Time on GPU: {gpu_time:.4f} seconds")

print(f"\nSpeedup (CPU time / GPU time): {cpu_time/gpu_time:.2f}x")
```
