```python
import torch
```

## Simple AutoGrad Useful in mathematical function

```python
x = torch.tensor(3.0, requires_grad=True)
y = x**2

y.backward()
x.grad
```

```python
x = torch.tensor(3.0, requires_grad = True)
y = x ** 2
z = torch.sin(y)

z.backward()
x.grad
```

# AutoGrad in Small Neural network

```python
def binary_cross_entropy(y, y_pred):
    epsilon = 1e-15  # Prevent log(0)
    y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)

    loss = -(y * torch.log(y_pred) +
             (1 - y) * torch.log(1 - y_pred))

    return torch.mean(loss)
```

```python
x = torch.tensor(6.7)
y = torch.tensor(0.0)

w = torch.tensor(1.0, requires_grad = True)
b = torch.tensor(0.0, requires_grad = True)

z = w * x + b

y_pred = torch.sigmoid(z)

loss = binary_cross_entropy(y, y_pred)

loss.backward()

print(w.grad)
print(b.grad)
```

## Vector Tensors

```python
x = torch.tensor([1.0, 2.0, 3.0], requires_grad = True)
y = (x**2).mean()
y.backward()
x.grad
```

```python
# clearing grad
x = torch.tensor(2.0, requires_grad = True)

y = x **2

y.backward()
print(x.grad)

x.grad.zero_()
```

## Disable gradient tracking

```python
# disable gradient tracking
x = torch.tensor(2.0, requires_grad = True)

y = x**2
y

y.backward()
x.grad
```

```python
#option1-- requires_grad_(False)
x.requires_grad_(False)

y = x **2
y
```

```python
#option2-- detach()
x = torch.tensor(2.0, requires_grad = True)

z = x.detach()
z

print(x**2)
print(z**2)
```

```python
#option3 -- torch.no_grad()
x = torch.tensor(2.0, requires_grad = True)

with torch.no_grad():
  y = x **2
y
```
